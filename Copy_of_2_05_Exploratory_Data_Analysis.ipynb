### Visualising the Data

Let's visualize the survival rate (mean survival) across different passenger classes (`Pclass`):

```python
# Calculate the mean survival rate for each passenger class
pclass_survival = titanic.groupby('Pclass')['Survived'].mean().reset_index()

# Create a bar chart of the mean survival rates
plt.bar(pclass_survival['Pclass'].astype(str), pclass_survival['Survived'], color=['#e74c3c', '#3498db', '#2ecc71'])
plt.xlabel('Passenger Class (Pclass)')
plt.ylabel('Mean Survival Rate')
plt.title('Mean Survival Rate by Passenger Class')
plt.show()
```

Interestingly, first-class passengers had a much higher survival rate than third-class passengers! Let's try another visualization: a scatter plot comparing age and fare to see if there is any relationship between how much passengers paid and their age.

```python
plt.scatter(titanic['Age'], titanic['Fare'], alpha=0.5, color='purple')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.title('Passenger Age vs Fare Paid')
plt.show()
```

As expected, there isn't a strong direct correlation between age and fare, though most higher fares were paid by adult passengers.

We may also want to see how the age distribution looks. Let's check with a distribution plot:

```python
ax = sns.displot(titanic['Age'].dropna(), kde=True, color='teal')
plt.title('Distribution of Passenger Ages')
plt.show()
```

The age distribution is somewhat right-skewed, showing a large proportion of young adults on board.

We can use the `pairplot()` method to visualize relationships between multiple numerical variables like `Age`, `Fare`, and `Pclass`:

```python
sns.pairplot(data=titanic[['Age', 'Fare', 'Pclass', 'Survived']].dropna(), hue='Survived', diag_kind='hist')
plt.show()
```

This pairwise grid helps us quickly spot patterns, such as survival distributions across different classes and ages.

We may also decide to remove a redundant or uninformative variable, such as `Ticket`:

```python
titanic = titanic.drop(['Ticket'], axis=1, errors='ignore')
titanic.head()
```
